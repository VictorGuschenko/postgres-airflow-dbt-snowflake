```python
data = requests.get('https://api.github.com/search/repositories?q=stars:%3E0&sort=stars&order=desc&per_page=10').json()
first_repo = data['items'][0]
first_repo
```

```python
table = [
    {
        'name': item['name'],
        'full_name': item['full_name'],
        'owner': item['owner']['login'],
        'stars': item['stargazers_count'],
        'forks': item['forks_count'],
        'open_issues': item['open_issues_count'],
        'language': item['language'],
        'url': item['html_url'],
    }
    for item in data['items']
]
```

```python
import pandas as pd

table = [
    {
        'name': item['name'],
        'full_name': item['full_name'],
        'owner': item['owner']['login'],
        'stars': item['stargazers_count'],
        'forks': item['forks_count'],
        'open_issues': item['open_issues_count'],
        'language': item.get('language') or 'Unknown',
        'url': item['html_url'],
    }
    for item in data['items']
]

df = pd.DataFrame(table)
df.to_csv('top_repos.csv', index=False)
```
